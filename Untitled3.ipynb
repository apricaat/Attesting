Интернет‑магазин тестирует новую версию карточки товара. Сейчас кнопка «Купить» находится внизу карточки, а в варианте B её переместили наверх — сразу после цены и описания. Цель — увеличить конверсию в покупку (доля пользователей, которые нажали «Купить» среди всех, кто открыл карточку).

Данные. За неделю собрали данные по двум группам:

Группа A (текущая версия): 10 000 пользователей, 850 покупок.
Группа B (новая версия): 10 200 пользователей, 950 покупок.
Задание:

Посчитай конверсию для каждой группы (в процентах).
Оцени, есть ли статистически значимое различие между группами (используй z‑тест для долей).
Напиши краткий вывод для менеджера: стоит ли внедрять новую версию и почему.

> Добавить блок с цитатой



```python
import math

# Данные
n_a = 10_000
x_a = 850
n_b = 10_200
x_b = 950

# Конверсии
p_a = x_a / n_a
p_b = x_b / n_b

# Общая доля (pooled proportion)
p_pool = (x_a + x_b) / (n_a + n_b)

# Стандартная ошибка разницы долей
se = math.sqrt(
    p_pool * (1 - p_pool) * (1/n_a + 1/n_b)
)

# Z-статистика
z = (p_b - p_a) / se

# Приближённое p-значение (двусторонний тест) через функцию ошибок erf
# Формула: p = 2 * (1 - Φ(|z|)), где Φ(z) ≈ 0.5 * (1 + erf(z / sqrt(2)))
p_value = 2 * (1 - 0.5 * (1 + math.erf(abs(z) / math.sqrt(2))))

print(f"Конверсия A: {p_a:.4f} ({p_a*100:.2f}%)")
print(f"Конверсия B: {p_b:.4f} ({p_b*100:.2f}%)")
print(f"Z-статистика: {z:.3f}")
print(f"P-значение: {p_value:.4f}")

if p_value < 0.05:
    print("Различие статистически значимо (p < 0.05). Рекомендуется внедрять вариант B.")
else:
    print("Различие не статистически значимо (p >= 0.05). Недостаточно оснований для внедрения.")
```
